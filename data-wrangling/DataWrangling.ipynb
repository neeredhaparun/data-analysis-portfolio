### 1. Load the Dataset


<h5>1.1 Import necessary libraries and load the dataset.</h5>


Ensure the dataset is loaded correctly by displaying the first few rows.


```python
# Import necessary libraries
import pandas as pd

# Load the Stack Overflow survey data
dataset_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"
df = pd.read_csv(dataset_url)

# Display the first few rows
print(df.head())
```

### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
print(df['Country'].unique())
print(f"\nCount of unique: {df['Country'].nunique()}\n")
print(df['Country'].value_counts())
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
country_mapping = {
    # Verbose official names → common names
    'United States of America': 'United States',
    'United Kingdom of Great Britain and Northern Ireland': 'United Kingdom',
    'Iran, Islamic Republic of...': 'Iran',
    'Venezuela, Bolivarian Republic of...': 'Venezuela',
    'Congo, Republic of the...': 'Congo',
    'Micronesia, Federated States of...': 'Micronesia',
    "Lao People's Democratic Republic": 'Laos',
    'Libyan Arab Jamahiriya': 'Libya',
    'Syrian Arab Republic': 'Syria',
    'United Republic of Tanzania': 'Tanzania',
    'Russian Federation': 'Russia',
    'Viet Nam': 'Vietnam',
    'Brunei Darussalam': 'Brunei',
    'Republic of Moldova': 'Moldova',
    'Republic of North Macedonia': 'North Macedonia',
    'Democratic Republic of the Congo': 'DR Congo',

    # Duplicate country pairs (different names, same country)
    "Democratic People's Republic of Korea": 'North Korea',
    'Republic of Korea': 'South Korea',

    # Outdated names → current official names
    'Swaziland': 'Eswatini',
    'Czech Republic': 'Czechia',
    'Cape Verde': 'Cabo Verde',
}

df['Country'] = df['Country'].replace(country_mapping)
# Handle irrelevant/non-country entries separately
df['Country'] = df['Country'].replace({'Nomadic': 'Unknown'})
# Handle missing values
df['Country'] = df['Country'].fillna('Unknown')

print(f"\nCount of unique: {df['Country'].nunique()}\n")
print(df['Country'].value_counts())
```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
df_emp_encoded = pd.get_dummies(df, columns = ['Employment'])
df_emp_encoded.head()
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
missing_values = df.isnull().sum().sort_values(ascending=False)
missing_values.head(10)
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
print(f"Missing values in ConvertedCompYearly before filling: {df['ConvertedCompYearly'].isnull().sum()}")
median = df['ConvertedCompYearly'].median()
df['ConvertedCompYearly'] = df['ConvertedCompYearly'].fillna(median)
print(f"Missing values in ConvertedCompYearly after filling: {df['ConvertedCompYearly'].isnull().sum()}")
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
print(f"Missing values in RemoteWork before filling: {df['RemoteWork'].isnull().sum()}")
most_freq_remote = df['RemoteWork'].mode()[0]
df['RemoteWork'] = df['RemoteWork'].fillna(most_freq_remote)
print(f"Missing values in RemoteWork after filling: {df['RemoteWork'].isnull().sum()}")
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
min_val = df['ConvertedCompYearly'].min()
max_val = df['ConvertedCompYearly'].max()
df['ConvertedCompYearly-MinMax'] = (df['ConvertedCompYearly'] - min_val) / (max_val - min_val)
df[['ConvertedCompYearly', 'ConvertedCompYearly-MinMax']].head()
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
import numpy as np
```

```python
df['ConvertedCompYearly-Log'] = np.log1p(df['ConvertedCompYearly'])
df[['ConvertedCompYearly', 'ConvertedCompYearly-Log']].head()
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
# Convert to numeric (handles text like "Less than 1 year" / "More than 50 years")
df['YearsCodePro_numeric'] = pd.to_numeric(
    df['YearsCodePro'].replace({'Less than 1 year': 0, 'More than 50 years': 51}), errors = 'coerce'
)
df['YearsCodePro_numeric'].unique()

def exp_level(years):
  if pd.isna(years):
    return 'Unknown'
  elif years <= 3:
    return 'Junior'
  elif years <= 8:
    return 'Mid-Level'
  else:
    return 'Senior'

df['ExperienceLevel'] = df['YearsCodePro_numeric'].apply(exp_level)
df[['YearsCodePro_numeric', 'ExperienceLevel']].head(10)
```
