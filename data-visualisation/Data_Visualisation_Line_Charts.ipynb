### Task 1: Trends in Compensation Over Age Groups


##### 1. Line Chart of Median `ConvertedCompYearly` by Age Group


- Track how the median yearly compensation (ConvertedCompYearly) changes across different age groups.

- Use a line chart to visualize these trends.


```python
df_comp_age = df[['Age', 'ConvertedCompYearly']].dropna()
df_comp_age_median = df_comp_age.groupby('Age', as_index=False)['ConvertedCompYearly'].median()
plt.figure(figsize=(12,7))
plt.plot(df_comp_age_median['Age'], df_comp_age_median['ConvertedCompYearly'], marker='o')
plt.title('Median Yearly Compensation by Age')
plt.xlabel('Age')
plt.ylabel('Median Yearly Compensation')
plt.xticks(rotation=45, ha='right')
plt.show()
```

##### 2. Line Chart of Median `ConvertedCompYearly` for Ages 25 to 45


For a closer look, plot a line chart focusing on the median compensation for respondents between ages 25 and 45.


```python
target_age = ['25-34 years old', '35-44 years old']
df_comp_age_filtered = df_comp_age[df_comp_age['Age'].isin(target_age)]

comp_age_median = df_comp_age_filtered.groupby('Age', as_index=False)['ConvertedCompYearly'].median()
plt.figure(figsize=(12,7))
plt.plot(comp_age_median['Age'], comp_age_median['ConvertedCompYearly'], marker='s')
plt.title('Median Yearly Compensation by Age')
plt.xlabel('Age')
plt.ylabel('Median Yearly Compensation')
plt.show()
```

### Task 2: Trends in Job Satisfaction by Experience Level



##### 1. Line Chart of Job Satisfaction (`JobSatPoints_6`) by Experience Level



- Use a column that approximates experience level to analyze how job satisfaction changes with experience.

- If needed, substitute an available experience-related column for `Experience`.


```python
df_exp_sat = df[['YearsCodePro', 'JobSatPoints_6']].dropna()

df_exp_sat['YearsCodePro_num'] = pd.to_numeric(
    df_exp_sat['YearsCodePro'].replace({
        'Less than 1 year': 0,
        'More than 50 years': 51
    }),
    errors='coerce')

df_exp_sat_med = df_exp_sat.groupby('YearsCodePro_num', as_index=False)['JobSatPoints_6'].median()
plt.figure(figsize=(12,7))
plt.plot(df_exp_sat_med['YearsCodePro_num'], df_exp_sat_med['JobSatPoints_6'], marker='o')
plt.title('Median Job Satisfaction by Years of Experience')
plt.xlabel('Years of Experience')
plt.ylabel('Median Job Satisfaction')
plt.show()
```

### Task 3: Trends in Job Satisfaction and Compensation by Experience


##### 1.Line Chart of Median ConvertedCompYearly Over Experience Level

- This line chart will track how median compensation (`ConvertedCompYearly`) changes with increasing experience.

- Use a column such as `WorkExp` or another relevant experience-related column.


```python
df_exp_comp = df[['WorkExp', 'ConvertedCompYearly']].dropna()

df_exp_comp_med = df_exp_comp.groupby('WorkExp', as_index=False)['ConvertedCompYearly'].median()

plt.figure(figsize=(12,7))
plt.plot(df_exp_comp_med['WorkExp'], df_exp_comp_med['ConvertedCompYearly'], marker='s')
plt.title('Median Yearly Compensation by Work Experience')
plt.xlabel('Work Experience')
plt.ylabel('Median Yearly Compensation')
plt.show()
```

##### 2.Line Chart of Job Satisfaction (`JobSatPoints_6`) Across Experience Levels

- Create a line chart to explore trends in job satisfaction (`JobSatPoints_6`) based on experience level.

- This chart will provide insight into how satisfaction correlates with experience over time


```python
df_exp_sat = df[['WorkExp', 'JobSatPoints_6']].dropna()
df_exp_sat_med = df_exp_sat.groupby('WorkExp', as_index=False)['JobSatPoints_6'].median()
df_exp_sat_mean = df_exp_sat.groupby('WorkExp', as_index=False)['JobSatPoints_6'].mean()
plt.figure(figsize=(14,7))
plt.plot(df_exp_sat_med['WorkExp'], df_exp_sat_med['JobSatPoints_6'], marker='o', label='Median')
plt.plot(df_exp_sat_mean['WorkExp'], df_exp_sat_mean['JobSatPoints_6'], marker='s', linestyle='--', label='Mean')

plt.title('Median Job Satisfaction by Work Experience (Mean vs Median)')
plt.xlabel('Work Experience')
plt.ylabel('Median Job Satisfaction')
plt.legend()
plt.show()


```
