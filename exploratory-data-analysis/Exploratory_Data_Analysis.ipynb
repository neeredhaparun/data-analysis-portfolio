#### Step 1: Install and Import Libraries


Install the necessary libraries for data manipulation and visualization.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

#### Step 2: Load and Preview the Dataset
Load the dataset from the provided URL. Use df.head() to display the first few rows to get an overview of the structure.


```python
# Load the Stack Overflow survey dataset
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)

# Set pandas option to display all columns
pd.set_option('display.max_columns', None)

# Display the first few rows of the dataset
df.head()
```

#### Step 3: Handling Missing Data


Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.


```python
print(df[['Employment', 'JobSat', 'RemoteWork']].isnull().sum())

# Remote work - filling missing values with most frequent value
most_freq_remote = df['RemoteWork'].mode()[0]
df['RemoteWork'] = df['RemoteWork'].fillna(most_freq_remote)

# JobSat - Job satisfaction - removing rows with missing Jobsat, since it's central to our analysis, and imputing it could distort results
df = df.dropna(subset = ['JobSat'])

print("Data after handling missing value\n",df[['Employment', 'JobSat', 'RemoteWork']].isnull().sum())
```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
df['YearsCodePro_numeric'] = pd.to_numeric(
    df['YearsCodePro'].replace({'Less than 1 year': 0, 'More than 50 years': 51}), errors = 'coerce'
)
df['YearsCodePro_numeric'].unique()
```

```python
bins = [0, 5, 10, 20, 100]
labels = ['0-5', '5-10', '10-20', '>20']
df['ExperienceRange'] = pd.cut(df['YearsCodePro_numeric'], bins=bins, labels=labels, include_lowest=True)
df[['YearsCodePro_numeric', 'ExperienceRange']].head(10)
```

```python
median_sat = df.groupby('ExperienceRange', observed=True)['JobSat'].median()
median_sat
```

```python
plt.figure(figsize=(12,6))
sns.barplot(x = median_sat.index, y = median_sat.values)
plt.xlabel('Experience Range')
plt.ylabel('Median Job Satisfaction')
plt.title('Median Job Satisfaction by Experience Range')
```

#### Step 5: Visualize Job Satisfaction


Use a count plot to show the distribution of `JobSat` values. This provides insights into the overall satisfaction levels of respondents.


```python
plt.figure(figsize=(12,6))
sns.countplot(data =df, x = 'JobSat')
plt.xlabel('Job Satisfaction')
plt.ylabel('Count')
plt.title('Distribution of Job Satisfaction')
```

#### Step 6: Analyzing Remote Work Preferences by Job Role


Analyze trends in remote work based on job roles. Use the `RemoteWork` and `Employment` columns to explore preferences and examine if specific job roles prefer remote work more than others.

- Use a count plot to show remote work distribution.

- Cross-tabulate remote work preferences by employment type (e.g., full-time, part-time) and job roles.


```python
plt.figure(figsize=(12,6))
sns.countplot(data = df, x = 'RemoteWork')
plt.xlabel('Remote Work')
plt.ylabel('Count')
plt.title('Distribution of Remote Work')
plt.show()
```

```python
cross_tab = pd.crosstab(df['Employment'], df['RemoteWork'])
cross_tab
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(cross_tab, annot=True, fmt='d', cmap='YlGnBu')

plt.title("Remote Work Preference by Employment")
plt.show()
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
top_countries = df['Country'].value_counts().head(5).index
df_top_countries = df[df['Country'].isin(top_countries)]

lang_by_country = df_top_countries[['Country', 'LanguageHaveWorkedWith']].dropna()
lang_by_country = lang_by_country.assign(
    LanguageHaveWorkedWith = lang_by_country['LanguageHaveWorkedWith'].str.split(';')
).explode('LanguageHaveWorkedWith')

top_languages = lang_by_country['LanguageHaveWorkedWith'].value_counts().head(5).index
lang_by_country_top = lang_by_country[lang_by_country['LanguageHaveWorkedWith'].isin(top_languages)]

lang_country_counts = pd.crosstab(lang_by_country_top['Country'], lang_by_country_top['LanguageHaveWorkedWith'])
print(lang_country_counts)

# Visualize with a heatmap
lang_country_counts.plot(kind='bar', figsize=(14, 7))
plt.title('Top Programming Languages by Country')
plt.xlabel('Country')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Language', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

#### Step 8: Correlation Between Experience and Satisfaction


Examine how years of experience (`YearsCodePro`) correlate with job satisfaction (`JobSatPoints_1`). Use a scatter plot to visualize this relationship.


```python
plt.figure(figsize=(12,6))
sns.scatterplot(df, x = 'YearsCodePro_numeric', y = 'JobSatPoints_1')
plt.xlabel('Years of Experience')
plt.ylabel('Job Satisfaction')
plt.title('Years of Experience vs. Job Satisfaction')
plt.show()
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
edu_employment = df[['EdLevel', 'Employment']].dropna()
edu_employment_crosstab = pd.crosstab(edu_employment['EdLevel'], edu_employment['Employment'])
plt.figure(figsize=(14, 10))
sns.heatmap(edu_employment_crosstab, fmt='d', cmap='coolwarm')
plt.title('Education Level vs Employment Type')
plt.xlabel('Employment Type')
plt.ylabel('Education Level')
plt.xticks(rotation=45, ha='right')
plt.show()
```
