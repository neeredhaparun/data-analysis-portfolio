### Step 1: Import Libraries and Load Data


- Import the `pandas`, `matplotlib.pyplot`, and `seaborn` libraries.


- You will begin with loading the dataset. You can use the pyfetch method if working on JupyterLite. Otherwise, you can use pandas' read_csv() function directly on their local machines or cloud environments.


```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Stack Overflow survey dataset
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)

# Display the first few rows of the dataset
df.head()
```

### Step 2: Examine the Structure of the Data


- Display the column names, data types, and summary information to understand the data structure.

- Objective: Gain insights into the dataset's shape and available variables.


```python
print('Shape\n', df.shape)
print('\nData types\n', df.dtypes)
print('\nSummary statistics\n', df.describe())
```

### Step 3: Handle Missing Data


- Identify missing values in the dataset.

- Impute or remove missing values as necessary to ensure data completeness.



```python
missing = df.isnull().sum()
# missing[['Employment', 'JobSat', 'YearsCodePro','RemoteWork', 'LanguageHaveWorkedWith', 'LanguageWantToWorkWith', 'EdLevel']]

for col in ['RemoteWork', 'EdLevel']:
  df[col] = df[col].fillna(df[col].mode()[0])

missing_after_filling = df.isnull().sum()
missing_after_filling[['RemoteWork', 'EdLevel']]
```

```python
#droping missing rows in JobSat and storing it in new df
df_jobsat_cleaned = df.dropna(subset = ['JobSat']).copy()
print('Missing in Jobsat: ', df_jobsat_cleaned['JobSat'].isnull().sum())
```

### Step 4: Analyze Key Columns


- Examine key columns such as `Employment`, `JobSat` (Job Satisfaction), and `YearsCodePro` (Professional Coding Experience).

- **Instruction**: Calculate the value counts for each column to understand the distribution of responses.



```python
print('Employement value count: \n', df['Employment'].value_counts())
print('\nJobSat value count: \n', df_jobsat_cleaned['JobSat'].value_counts())
print('\nYearsCodePro value count: \n', df['YearsCodePro'].value_counts())
```

### Step 5: Visualize Job Satisfaction (Focus on JobSat)


- Create a pie chart or KDE plot to visualize the distribution of `JobSat`.

- Provide an interpretation of the plot, highlighting key trends in job satisfaction.


```python
plt.figure(figsize = (12,6))
df_jobsat_cleaned['JobSat'].value_counts().plot(kind = 'pie', autopct = '%1.1f%%')
plt.title('Job Satisfaction Distribution')
plt.show()
```

### Step 6: Programming Languages Analysis


- Compare the frequency of programming languages in `LanguageHaveWorkedWith` and `LanguageWantToWorkWith`.
  
- Visualize the overlap or differences using a Venn diagram or a grouped bar chart.


```python
have_worked = df['LanguageHaveWorkedWith'].dropna().str.split(';').explode()
want_to_work = df['LanguageWantToWorkWith'].dropna().str.split(';').explode()

have_worked_counts = have_worked.value_counts()
want_to_work_counts = want_to_work.value_counts()

comparison_df = pd.DataFrame({
    'HaveWorkedWith': have_worked_counts,
    'WantToWorkWith': want_to_work_counts
}).fillna(0)

comparison_df.plot(kind = 'bar', figsize=(20,10))
plt.title('Comaprison of HaveWorkedWith and WantToWorkWith')
plt.xlabel('Programming Languages')
plt.ylabel('Count')
plt.xticks(rotation = 45)
plt.legend(title = 'Category')
plt.show()
```

### Step 7: Analyze Remote Work Trends


- Visualize the distribution of RemoteWork by region using a grouped bar chart or heatmap.


```python
top_countries = df['Country'].value_counts().head(10).index
df_top = df[df['Country'].isin(top_countries)]

df_remote_dist = pd.crosstab(df_top['Country'], df_top['RemoteWork'])
df_remote_dist.plot(kind='bar', figsize=(14,7))
plt.title('Remote Work Distribution by Region')
plt.xlabel('Region')
plt.ylabel('Count')
plt.xticks(rotation = 45)
plt.show()
```

### Step 8: Correlation between Job Satisfaction and Experience


- Analyze the correlation between overall job satisfaction (`JobSat`) and `YearsCodePro`.
  
- Calculate the Pearson or Spearman correlation coefficient.


```python
import scipy.stats as stats
```

```python
df_jobsat_cleaned['YearsCodePro_numeric'] = pd.to_numeric(df_jobsat_cleaned['YearsCodePro'].replace({
    'Less than 1 year': 0,
    'More than 50 years': 51
}), errors = 'coerce')

df_jobsat_cleaned.dropna(subset = ['JobSat', 'YearsCodePro_numeric'], inplace = True)

coef, p_value = stats.pearsonr(df_jobsat_cleaned['JobSat'], df_jobsat_cleaned['YearsCodePro_numeric'])
print('Pearson Correlation Coefficient: ', coef)
print('P-value: ', p_value)
```

### Step 9: Cross-tabulation Analysis (Employment vs. Education Level)


- Analyze the relationship between employment status (`Employment`) and education level (`EdLevel`).

- **Instruction**: Create a cross-tabulation using `pd.crosstab()` and visualize it with a stacked bar plot if possible.


```python
emp_edu = df[['Employment', 'EdLevel']]
emp_edu = emp_edu.assign(
    Employment=emp_edu['Employment'].str.split(';')
).explode('Employment')

emp_edu_crosstab = pd.crosstab(emp_edu['Employment'], emp_edu['EdLevel'])
emp_edu_crosstab


emp_edu_crosstab.plot(kind='bar', stacked=True, figsize=(14, 8))
plt.title('Employment Status by Education Level')
plt.xlabel('Employment Status')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Education Level', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
